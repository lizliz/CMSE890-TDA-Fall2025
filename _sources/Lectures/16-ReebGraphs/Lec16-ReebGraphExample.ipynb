To see some examples, we will use the [ceREEBerus](https://munchlab.github.io/ceREEBerus/) package. 

```python
# Run in case you don't already have it installed:
# pip install cereeberus
# Note you need at least version 0.1.9 of ceREEBerus due to updates in naming conventions.
```

```python
from cereeberus import ReebGraph, LowerStar, computeReeb
from cereeberus.data.ex_torus import Torus
```

The example we've been using in the class lecture can be coded as follows. Notice that because my triangles in the code are backwards from the figures from class, this is actually a mirror image of that representation. 

```python
T = Torus()
T.generate_grid(grid_size = 4)
# f = {3:11, 2:13, 1:15, 0:7, 
#      7:14, 6:1, 5:3, 4:16, 
#      11:9, 10:2, 9:4, 8:8, 
#      15:10, 14:12, 13:6, 12:5}
# for v in f.keys():
#     T.assign_filtration(v, f[v])
T.assign_random_values(1,100)
T.draw()
```

Then to compute the Reeb graph of this torus, we have the following. 

```python
R = computeReeb(T, verbose = False)
R.remove_all_regular_vertices()
R.draw(cpx = 3)
```
